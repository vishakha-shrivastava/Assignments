Flexible Arguments

```python
#function to shopping cart details
#customer_name is a required argument
#*product allows the function to accept any number of arguments
def shopping_cart(customer_name, *products):
  #display the customer name
  print("customer name:", customer_name)
  #display each product purchased by the customer
  print("products purchased:")
  for product in products:
    print("-", product)
#display the total number of products purchased
    print("Number of products:",len(products))
#calling the function with one product
shopping_cart("Vishakha","Laptop")

print() #add the blank line between the outputs

#Calling the function with multiple products
shopping_cart("Vishakha", "Laptop", "Mobile", "Tablet")



```

Map


```python
ages = (18, 25, 32, 16, 40) # Create a tuple containing different ages
# Use map() to add 1 to each age using a lambda function
result = map(lambda age: age + 1 , ages)
# Convert the map object into a list and print the result
print(list(result))
```

Filter

```python
ages = (18, 25, 32, 16, 40) # Create a tuple containing different ages

# Keep only the ages that are 18 or older
result = filter(lambda age: age >= 18, ages)

# Convert the filter object into a list
print(list(result))
```

Reduce

```python
from functools import reduce
# Create a tuple containing different ages
ages = (18, 25, 32, 16, 40)
# Use reduce() to add all the ages together
result = reduce(lambda x, y: x + y, ages)
#print the result
print(result)
```
